```python
# CRUD (1 балл) - реализуем свою базу данных с помощью DDL-запросов

cursor.execute("DROP DATABASE IF EXISTS lexicon") # удаляем старую базу, если она существует

cursor.execute("CREATE DATABASE IF NOT EXISTS lexicon CHARACTER SET utf8mb4")
cursor.execute("USE lexicon")

ddl = [
"""
CREATE TABLE IF NOT EXISTS Lemma (
    id INT AUTO_INCREMENT PRIMARY KEY,
    lemma VARCHAR(30) NOT NULL,
    homonym_number INT NOT NULL DEFAULT 1,
    part_of_speech VARCHAR(8),
    v_stem VARCHAR(3),
    morphology VARCHAR(200),
    etymology TEXT,
    dialect VARCHAR(100),
    stylistic VARCHAR(100),
    UNIQUE KEY uq_lemma_homonym (lemma, homonym_number)
)
""",
"""
CREATE TABLE IF NOT EXISTS Meaning_polysemy (
    id INT AUTO_INCREMENT PRIMARY KEY,
    lemma_id INT NOT NULL,
    meaning TEXT NOT NULL,
    example TEXT,
    FOREIGN KEY (lemma_id) REFERENCES Lemma(id)
        ON DELETE CASCADE ON UPDATE CASCADE
)
""",
"""
CREATE TABLE IF NOT EXISTS Speaker (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(100),
    gender ENUM('M','F'),
    date_of_birth DATE,
    location TEXT
)
""",
"""
CREATE TABLE IF NOT EXISTS Audio (
    id INT AUTO_INCREMENT PRIMARY KEY,
    lemma_id INT NOT NULL,
    speaker_id INT,
    file_path VARCHAR(200) NOT NULL,
    date_of_recording DATE,
    FOREIGN KEY (lemma_id) REFERENCES Lemma(id)
        ON DELETE CASCADE ON UPDATE CASCADE,
    FOREIGN KEY (speaker_id) REFERENCES Speaker(id)
        ON DELETE SET NULL ON UPDATE CASCADE
)
"""
]

for stmt in ddl:
    cursor.execute(stmt)
```

```python
# Запросы (2 балла) - запрос 1

# выводит топ-10 лемм с наибольшим числом соответствующих аудио и считает,
# сколько аудиозаписей для этих лемм существует и сколько различных дикторов их записывали, а также выводит дату последней записи

query1 = """
SELECT
  L.lemma,
  COUNT(A.id) AS recordings_count,
  COUNT(DISTINCT A.speaker_id) AS distinct_speakers,
  MAX(A.date_of_recording) AS last_recording_date
FROM Lemma L
LEFT JOIN Audio A ON A.lemma_id = L.id
GROUP BY L.lemma
ORDER BY recordings_count DESC
LIMIT 10;
"""

cursor.execute(query1)
rows = cursor.fetchall()
rows
```

```python
# Запросы (2 балла) - запрос 2

# для каждого диктора возвращается топ-2 наиболее часто записанных им лемм (если диктор записывал 0-1 лемму, то запрос их вернёт)

query2 = """
WITH speaker_lemma_counts AS (
    SELECT
        S.id AS speaker_id,
        S.name AS speaker_name,
        L.lemma,
        COUNT(A.id) AS recordings_by_speaker
    FROM Speaker S
    LEFT JOIN Audio A ON A.speaker_id = S.id
    LEFT JOIN Lemma L ON L.id = A.lemma_id
    GROUP BY S.id, S.name, L.lemma
)
SELECT *
FROM (
    SELECT *,
        ROW_NUMBER() OVER (
            PARTITION BY speaker_id
            ORDER BY recordings_by_speaker DESC
        ) AS rn
    FROM speaker_lemma_counts
) t
WHERE rn <= 2
ORDER BY speaker_id, rn;
"""

cursor.execute(query2)
rows = cursor.fetchall()
rows
```

```python
# Программирование (2 балла) - пользовательская процедура и пример её вызова

# добавляет новую аудиозапись в таблицу Audio и возвращает её ID

cursor.execute("DROP PROCEDURE IF EXISTS add_new_audio")

cursor.execute("""
CREATE PROCEDURE add_new_audio(
  IN p_lemma_id INT,
  IN p_speaker_id INT,
  IN p_file VARCHAR(200),
  IN p_date DATE,
  OUT p_new_id INT
)
BEGIN
  INSERT INTO Audio (lemma_id, speaker_id, file_path, date_of_recording)
  VALUES (p_lemma_id, p_speaker_id, p_file, p_date);
  SET p_new_id = LAST_INSERT_ID();
END
""")

args = [5, 1, "/audio/test.mp3", "2023-08-01", 0]
result_args = cursor.callproc("add_new_audio", args)

print("Inserted audio id:", result_args[-1])
```
